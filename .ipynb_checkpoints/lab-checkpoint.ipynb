#  The code for bringing the words from the csv word by using its unique code 

```python
import pandas as pd


def get_word_by_unique_code(unique_code_to_find, csv_filename="words.csv"):
  """Searches the CSV file for a given unique code and returns a dictionary

  containing all the values for that row.
  """
  try:
    df = pd.read_csv(csv_filename)
  except FileNotFoundError:
    print(
        f"Error: The file '{csv_filename}' was not found. Please check the"
        " path."
    )
    return None

  # Clean up column names in case of trailing spaces
  df.columns = df.columns.str.strip()

  # Filter the DataFrame for the matching unique code
  # We convert both to string to avoid type mismatch issues (e.g., int vs str)
  match = df[df["unigue code"].astype(str) == str(unique_code_to_find)]

  if match.empty:
    print(f"No entry found with Unique Code: {unique_code_to_find}")
    return None

  # Extract the first matching row as a dictionary
  row_data = match.iloc[0].to_dict()
  return row_data


# Example usage:
if __name__ == "__main__":
  # Test with unique code 31 (tsuki)
  code_to_search = 31
  result = get_word_by_unique_code(code_to_search)

  if result:
    print(f"--- Data for Unique Code: {code_to_search} ---")
    for key, value in result.items():
      print(f"{key}: {value}")
```

# the function that give us the new words unqie codes 

```python
import pandas as pd


def get_and_update_all_unprocessed_codes(csv_filename="words.csv"):
  """Searches for all rows where 'added' is 0, retrieves their unique codes,

  updates their 'added' values to 1 in the CSV, and returns a list of the unique
  codes.
  Returns an empty list if all rows have already been added.
  """
  try:
    df = pd.read_csv(csv_filename)
  except FileNotFoundError:
    print(
        f"Error: The file '{csv_filename}' was not found. Please check the"
        " path."
    )
    return []

  # Clean up column names in case of trailing spaces
  df.columns = df.columns.str.strip()

  # Ensure the 'added' column exists, if not initialize it with 0
  if "added" not in df.columns:
    df["added"] = 0

  # Find all rows where 'added' is equal to 0
  unprocessed_rows = df[df["added"] == 0]

  if unprocessed_rows.empty:
    print("All rows have already been processed (added = 1).")
    return []

  # Retrieve all unique codes from the unprocessed rows
  unique_codes = unprocessed_rows["unigue code"].tolist()

  # Update the 'added' value to 1 for all matching rows
  df.loc[df["added"] == 0, "added"] = 1

  # Save the updated DataFrame back to the CSV file
  df.to_csv(csv_filename, index=False)

  print(
      f"Processed and updated {len(unique_codes)} rows. Unique Codes:"
      f" {unique_codes}"
  )
  return unique_codes


# Example usage:
if __name__ == "__main__":
  codes = get_and_update_all_unprocessed_codes("words.csv")
  print("Returned Unique Codes list:", codes)
```
