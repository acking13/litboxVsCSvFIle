#  The code for bringing the words from the csv word by using its unique code 

```python
import pandas as pd


def get_word_by_unique_code(unique_code_to_find, csv_filename="words.csv"):
  """Searches the CSV file for a given unique code and returns a dictionary

  containing all the values for that row.
  """
  try:
    df = pd.read_csv(csv_filename)
  except FileNotFoundError:
    print(
        f"Error: The file '{csv_filename}' was not found. Please check the"
        " path."
    )
    return None

  # Clean up column names in case of trailing spaces
  df.columns = df.columns.str.strip()

  # Filter the DataFrame for the matching unique code
  # We convert both to string to avoid type mismatch issues (e.g., int vs str)
  match = df[df["unigue code"].astype(str) == str(unique_code_to_find)]

  if match.empty:
    print(f"No entry found with Unique Code: {unique_code_to_find}")
    return None

  # Extract the first matching row as a dictionary
  row_data = match.iloc[0].to_dict()
  return row_data


# Example usage:
if __name__ == "__main__":
  # Test with unique code 31 (tsuki)
  code_to_search = 31
  result = get_word_by_unique_code(code_to_search)

  if result:
    print(f"--- Data for Unique Code: {code_to_search} ---")
    for key, value in result.items():
      print(f"{key}: {value}")
```
